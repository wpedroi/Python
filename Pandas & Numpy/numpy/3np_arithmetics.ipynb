```python
import numpy as np

# Scalar Arithmetic
```

```python
array = np.array([1, 2, 3])
```

```python
# basic Scalar arithmetic

print(array + 1)
print(array - 2)
print(array * 3)
print(array / 4)
print(array ** 5)
```

```python
# vectorized math funcs

print(np.sqrt(array))
print(np.round(array))
print(np.pi)

# Exercise

raddi = array
print(np.pi * raddi ** 2)

#  Element-wise arithmetic

array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

print(array1 + array2)
print(array1 - array2)
print(array1 * array2)
print(array1 / array2)
print(array1 ** array2)

# Comparison Operators

scores = np.array([91, 55, 100, 73, 82, 64])
# print(scores < 60)

scores[scores < 60] = 0
print(scores)
```
