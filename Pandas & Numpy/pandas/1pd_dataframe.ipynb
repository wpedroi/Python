```python
import pandas as pd
```

```python
data = {
    "Name":["Ahmed", "Husain", "Adham"],
    "Age":[30, 20, 21]
}

df = pd.DataFrame(data, index=["Employee 1", "Employee 2", "Employee 3"])
df
```

```python
print(df.loc["Employee 1"])

print(df.iloc[2])
```

```python
# to add a column
df["Job"] = ["Cook", "N/A", "Cashier"]
df
```

```python
# to add a row
new_row = pd.DataFrame([{"Name": "Mohammed", "Age": 28, "Job": "Engineer"},
                        {"Name": "Ali", "Age": 29, "Job": "Doctor"}], index=["Employee 4", "Employee 5"])
df = pd.concat([df, new_row])
df
```

```python
df = df.drop("Employee 5")
df
```
