```python
import pandas as pd
```

```python
df = pd.read_csv(r"continents2.csv")
df
```

```python
# Whole dataframe

# mean aggregate function
display(df.mean(numeric_only=True))
# sum aggregate function
display(df.sum(numeric_only=True))
# min aggregate function
display(df.min(numeric_only=True))
# max aggregate function
display(df.max(numeric_only=True))
# count aggregate function
display(df.count(numeric_only=True))
```

```python
# mean aggregate function
print(df["country-code"].mean(numeric_only=True))
```
