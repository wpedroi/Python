```python
import pandas as pd
```

```python
# data cleaning  = The process of fixing or removing:
#                  incomplete, incorrect , or irrelevant data.
#                  ~75% of workdone with pandas is data cleaning
df = pd.read_csv(r"continents2.csv")
df
```

```python
# drop irrelevent columns
dd = df.drop(columns=['alpha-3'])
dd
```

```python
# handle missing data
mm = df.dropna(subset=['region'])
mm
```

```python
# fill missing values
fill = df.fillna({"region": "None"})
fill
```

```python
# fix inconsistent values
df["region"] = df["region"].replace({"ASIA": "Asia"})
df
```

```python
#  Standarize text lower
df["name"] = df["name"].str.lower()
df
#  Standarize text upper
df["name"] = df["name"].str.upper()
df
#  Standarize text strip
df["name"] = df["name"].str.strip()
df
```

```python
# fix data types
df["country-code"] = df["country-code"].astype(object)
df.info()
```

```python
# fix duplicate values
df = df.drop_duplicates()
df
```
